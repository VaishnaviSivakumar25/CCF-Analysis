```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("creditcard.csv")
```

```python
#droping unwanted data

data.drop(["Time"],axis =1,inplace=True)
```

```python
#log-Transformation
data["LogAmount"] = np.log1p(data['Amount'])
```

```python
#dropping original Amountcolumn we'have transformed

data.drop(['Amount'],axis =1,inplace=True)
```

```python
#visualize of inputs
plt.figure(figsize=(10,6))
sns.histplot(data["LogAmount"],bins=100,kde = True)
#kde is used to find out the skewness by using curevelines
plt.title("Distribution of Log-Transformed Amount")
plt.show()
```
